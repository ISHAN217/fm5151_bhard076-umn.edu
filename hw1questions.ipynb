# Homework 1 — Mortgage questions

Loan: initial balance **$500,000**, APR **7%**, term **30 years**. Calculations keep full Python precision; displayed dollar amounts are rounded for readability.


```python
from mortgage import amortize
from root import bisection

initial_balance = 500_000
apr = 0.07
years = 30
schedule = amortize(apr, years, initial_balance)
level_payment = schedule["pmt"][0]
```

## 1. Level monthly payment

**Answer:** $3,326.51 per month (full precision: $3,326.5124758959).


```python
print(f"${level_payment:,.2f}")
```

## 2. Total principal repaid

**Answer:** $500,000.00. A fully repaid loan returns the original principal.


```python
total_principal = sum(schedule["principal_repaid"])
print(f"${total_principal:,.2f}")
```

## 3. Total interest paid

**Answer:** $697,544.49.


```python
total_interest = sum(schedule["interest"])
print(f"${total_interest:,.2f}")
```

## 4. Interest or principal at the beginning?

**Answer:** More goes to **interest** in month 1: $2,916.67 interest versus $409.85 principal.


```python
print(f"Month 1 interest: ${schedule['interest'][0]:,.2f}")
print(f"Month 1 principal: ${schedule['principal_repaid'][0]:,.2f}")
```

## 5. Does that ratio change?

**Answer:** Yes. As the balance declines, interest falls and more of each level payment repays principal. In the final month, $19.29 goes to interest and $3,307.22 to principal.


```python
print(f"Final month interest: ${schedule['interest'][-1]:,.2f}")
print(f"Final month principal: ${schedule['principal_repaid'][-1]:,.2f}")
```

## 6. Pay an extra $50 each month

**Answer:** The loan is paid off in **month 343**. The required final payment that month is **$2,616.15**.


```python
faster = amortize(apr, years, initial_balance, level_payment + 50)
print(f"Payoff month: {faster['period'][-1]}")
print(f"Final payment: ${faster['pmt'][-1]:,.2f}")
```

## 7. Pay $50 less each ordinary month

**Answer:** To clear the remaining balance at the end of month 360, the final payment is **$64,275.06**. This is a balloon payment.


```python
slower = amortize(apr, years, initial_balance, level_payment - 50)
print(f"Final month: {slower['period'][-1]}")
print(f"Final payment: ${slower['pmt'][-1]:,.2f}")
```

## 8. Payments for target total interest

Bisection solves `total_interest(payment) - target = 0`. The payments below show six decimals; the calculations retain full precision. Rounding these payments to cents changes the resulting total interest.

| Target total interest | Monthly payment |
|---:|---:|
| $100,000 | $9,426.490248 |
| $250,000 | $5,029.031481 |
| $500,000 | $3,667.754095 |
| $750,000 | $3,265.515641 |
| $1,000,000 | $2,974.808162 |
| $2,000,000 | $1,811.978247 |


```python
def total_interest_for(payment):
    projection = amortize(apr, years, initial_balance, payment)
    return sum(projection["interest"])

targets = [100_000, 250_000, 500_000, 750_000, 1_000_000, 2_000_000]
upper_payment = initial_balance * (1 + apr / 12)

for target in targets:
    def error(payment):
        return total_interest_for(payment) - target
    payment = bisection(error, 0, upper_payment, tol=1e-10, max_iters=100)
    print(f"Target ${target:>9,.0f}: monthly payment ${payment:,.6f}")
```
